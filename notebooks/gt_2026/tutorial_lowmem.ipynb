### Memory budget

This notebook is tuned to stay **under 16 GiB of RAM**. It works with several models that do not fit in memory at the same time, so it loads each one, uses it, and then gives the memory back before moving on. The second cell below defines the helpers that do this (`mem`, `load_hf_backend`, `unload_hf_backend`, `ollama_stop`, `free_all_models`), and every module calls `mem()` so you can watch the budget as you go.

Measured peaks per module on an Apple Silicon laptop:

| Module | Resident models | Peak |
|--------|-----------------|------|
| 1–2 | ollama granite4.1:3b, then HF granite-4.0-micro | ~7.5 GiB |
| 3 | HF granite-3.3-2b + stembolts adapter | ~6 GiB |
| 4 (extract) | docling + ollama granite4.1:3b | ~7.5 GiB |
| 4 (pricing) | HF granite-4.0-micro + ollama granite4.1:3b | ~12.5 GiB |

## Keeping the notebook inside 16 GiB

The next cell defines the memory helpers described above. Run it now — the rest of the notebook depends on it.

The short version: **load a model, use it, give the memory back.** `load_hf_backend()` enforces that only one Hugging Face model is resident at a time, `unload_hf_backend()` returns the weights (including adapter weights and cached KV state), and `ollama_stop()` evicts models from the ollama server. `mem()` prints where you stand at any point.

```python
# ===========================================================================
# Memory budget + model lifecycle helpers.
#
# This notebook is built to run in ~16 GiB of RAM. Three things blow that
# budget if left alone:
#
#  1. Hugging Face weights. A `LocalHFBackend` pins its model for as long as
#     the object is alive, and this notebook uses three of them across four
#     modules: granite-4.0-micro (6.4 GiB), granite-3.3-2b (4.7 GiB), then
#     granite-4.0-micro again. Left to stack up, that is ~19 GiB of weights.
#  2. Ollama's KV cache, which is sized from `num_ctx`. At the default of
#     131072, granite4.1:3b is resident at 13 GB; at 16384 the same model is
#     resident at 3.6 GB. Nothing here needs more than ~4k tokens of context.
#  3. Docling's OCR models, which permanently add ~1.9 GiB to this kernel and
#     extract the same text as no-OCR on the documents used here (`DO_OCR`).
#
# The helpers below handle (1) and (2):
#   mem(label)             print where you stand
#   load_hf_backend(id)    get an HF backend, keeping at most ONE resident
#   unload_hf_backend(id)  give the weights back (all backends if no id)
#   ollama_stop(*models)   evict models from the ollama server's memory
#   reclaim(label)         run a GC + allocator sweep
#   free_all_models()      all of the above
# ===========================================================================
import gc
import os
import subprocess

import psutil
import torch

from mellea.backends.model_options import ModelOption

MEM_VERBOSE = True

# Cap ollama's context window, and so its KV cache. The largest prompt in this
# notebook is a product catalog used as grounding context, at ~4k tokens.
OLLAMA_NUM_CTX = 16384
OLLAMA_MODEL_OPTS = {ModelOption.CONTEXT_WINDOW: OLLAMA_NUM_CTX}

# Docling's OCR stack costs ~1.9 GiB that is never returned. Every document in
# this tutorial has a real text layer, so OCR gains nothing. Set this to True if
# you swap in scanned documents.
DO_OCR = False


# --- reporting -------------------------------------------------------------
def _accelerator() -> str:
    if torch.cuda.is_available():
        return "cuda"
    if torch.backends.mps.is_available():
        return "mps"
    return "cpu"


def _device_allocated_gib() -> float:
    """GiB the torch allocator holds on the accelerator.

    On Apple Silicon, model weights live in Metal's shared memory pool and do
    NOT show up in the process's RSS, so RSS alone badly understates a loaded
    model -- a 6.4 GiB model moves RSS by less than 200 MiB.
    """
    try:
        if _accelerator() == "cuda":
            return torch.cuda.memory_reserved() / 2**30
        if _accelerator() == "mps":
            return torch.mps.driver_allocated_memory() / 2**30
    except Exception:
        pass
    return 0.0


def mem(label: str = "") -> float:
    """Print and return the memory this kernel is responsible for, in GiB.

    That is RSS + accelerator allocations. System-wide usage is printed too,
    because ollama holds its models in a separate process.
    """
    rss = psutil.Process(os.getpid()).memory_info().rss / 2**30
    dev = _device_allocated_gib()
    vm = psutil.virtual_memory()
    if MEM_VERBOSE:
        print(
            f"[mem] {label:<34} kernel {rss + dev:5.2f} GiB "
            f"(rss {rss:4.2f} + {_accelerator()} {dev:4.2f}) | "
            f"system {(vm.total - vm.available) / 2**30:5.2f}/{vm.total / 2**30:.0f} GiB"
        )
    return rss + dev


def reclaim(label: str = "after reclaim") -> float:
    """Collect garbage and return pooled allocator memory to the OS/device."""
    gc.collect()
    if _accelerator() == "cuda":
        torch.cuda.empty_cache()
    elif _accelerator() == "mps":
        torch.mps.empty_cache()
    gc.collect()
    return mem(label)


# --- ollama ----------------------------------------------------------------
def ollama_running() -> list[str]:
    """Names of the models ollama currently holds in memory."""
    try:
        out = subprocess.run(
            ["ollama", "ps"], capture_output=True, text=True, timeout=15, check=True
        ).stdout
    except Exception:
        return []
    return [line.split()[0] for line in out.splitlines()[1:] if line.strip()]


def ollama_stop(*models: str) -> None:
    """Evict models from the ollama server's memory; with no args, evict all."""
    for name in models or ollama_running():
        subprocess.run(["ollama", "stop", name], capture_output=True, timeout=60)
        if MEM_VERBOSE:
            print(f"[mem] ollama stop {name}")


# --- huggingface -----------------------------------------------------------
_HF_BACKENDS: dict = {}


def _hf_key(model_id) -> str:
    name = getattr(model_id, "hf_model_name", None)
    return str(name if name is not None else model_id)


def _drop_notebook_refs(target) -> list[str]:
    """Delete notebook names bound to `target` so it can actually be collected.

    A backend stays alive as long as anything points at it, including the
    variable you assigned it to, a MelleaSession wrapping it, and IPython's
    `Out`/`_` cache of previously displayed values.
    """
    released: list[str] = []
    try:
        ip = get_ipython()  # noqa: F821
    except NameError:
        return released
    if ip is None:
        return released

    ns = ip.user_ns
    for name, val in list(ns.items()):
        if name.startswith("_") or name in ("In", "Out"):
            continue
        try:
            hit = val is target or getattr(val, "backend", None) is target
        except Exception:
            hit = False
        if hit:
            del ns[name]
            released.append(name)

    ip.displayhook.flush()  # clears Out / _N
    for name in ("_", "__", "___"):
        ns.pop(name, None)
    return released


def unload_hf_backend(model_id=None) -> None:
    """Free a LocalHFBackend's weights, adapters and KV caches.

    With no argument, unloads every backend created via `load_hf_backend`.
    """
    from mellea.backends.huggingface import LocalHFBackend

    keys = list(_HF_BACKENDS) if model_id is None else [_hf_key(model_id)]
    for key in keys:
        backend = _HF_BACKENDS.pop(key, None)
        if backend is None:
            continue

        # 1. Detach any LoRA/aLoRA adapter weights from the base model.
        for qualified_name in list(backend.list_adapters()):
            try:
                backend.unload_adapter(qualified_name)
            except Exception as e:
                print(f"[mem] could not unload adapter {qualified_name}: {e}")

        # 2. Break the adapter <-> backend reference cycle. An adapter holds
        #    `adapter.backend` and the backend holds `_added_adapters`, so
        #    neither is freed on refcount alone.
        for adapter in list(backend._added_adapters.values()):
            adapter.backend = None
        backend._added_adapters.clear()
        backend._loaded_adapters.clear()

        # 3. Drop cached KV state. `_cached_blocks` is a *class* attribute, so
        #    it outlives any single backend and must be cleared explicitly.
        LocalHFBackend._cached_blocks.clear()
        cache = getattr(backend, "_cache", None)
        if cache is not None and hasattr(cache, "cache"):
            cache.cache.clear()

        # 4. Drop the model and tokenizers, then any notebook name pointing at
        #    the backend, then sweep.
        for attr in ("_model", "_tokenizer", "_llguidance_tokenizer", "formatter"):
            try:
                delattr(backend, attr)
            except AttributeError:
                pass
        released = _drop_notebook_refs(backend)
        del backend

        if MEM_VERBOSE:
            extra = f"; released {', '.join(released)}" if released else ""
            print(f"[mem] unloaded HF model {key}{extra}")
        reclaim(f"after unload {key}")


def load_hf_backend(model_id, **kwargs):
    """Get a LocalHFBackend, keeping at most one HF model resident.

    Reuses an already-loaded backend for the same model, and unloads any
    backend for a *different* model first, so HF weights never stack up.
    """
    from mellea.backends.huggingface import LocalHFBackend

    key = _hf_key(model_id)
    for other in [k for k in _HF_BACKENDS if k != key]:
        unload_hf_backend(other)
    if key not in _HF_BACKENDS:
        mem(f"before loading {key}")
        _HF_BACKENDS[key] = LocalHFBackend(model_id=model_id, **kwargs)
        mem(f"loaded {key}")
    return _HF_BACKENDS[key]


def free_all_models() -> float:
    """Release everything: HF weights in this kernel and models held by ollama."""
    unload_hf_backend()
    ollama_stop()
    return reclaim("after freeing all models")


mem("baseline")
```

# Module 1: Hello, Mellea!

Running `mellea.start_session()` initialize a new `MelleaSession`. The session holds three things:
1. The model to use for this session. In this tutorial we will use granite4.1:3b.
2. An inference engine; i.e., the code that actually calls our model. We will be using ollama, but you can also use Huggingface or any OpenAI-compatible endpoint.
3. A `Context`, which tells Mellea how to remember context between requests. This is sometimes called the "Message History" in other frameworks. Throughout this tutorial, we will be using a `SimpleContext`. In `SimpleContext`s, **every request starts with a fresh context**. There is no preserved chat history between requests. Mellea provides other types of context, but today we will not be using those features. See the Tutorial for further details.

```python
import mellea

# OLLAMA_MODEL_OPTS caps num_ctx. Ollama sizes its KV cache from num_ctx, and at
# the 131072 default this 3B model is resident at 13 GB rather than 3.6 GB.
m = mellea.start_session(model_options=OLLAMA_MODEL_OPTS)

answer = m.chat(
    "tell me some fun trivia about IBM and the early history of AI."
)
print(answer.content)
```

# Module 2: Instruct-Validate-Repair

Instruct-Validate-Repair is a design pattern for building robust automation using LLMs. The idea is simple:
1. Instruct the model to perform a task and specify requirements on the output of the task.
2. Validate that these requirements are satisfied by the model's output.
3. If any requirements fail, try to repair.

| Type | Class | Validated by | Cost |
|------|-------|-------------|------|
| **Programmatic** | `req(..., validation_fn=simple_validate(fn))` | A `str → bool` function you write | Free / cost of Python |
| **LLM-as-judge** | `LLMaJRequirement(description)` | The model itself, given the description | One extra inference call |
| **Granite adapter functions** | `check_answerability`, `check_context_relevance` (from `mellea.stdlib.components.intrinsic`) | Hugging Face transformers backend | ~5-10% latency overhead (requires `mellea[hf]`) |

### Step 1: Extracting a Bill of Materials from Construction Documents

We'll start by parsing the tables in the construction plan to build a bill of materials.

```python
import logging
logging.getLogger("mellea").setLevel(logging.ERROR)

from mellea.stdlib.components.docs.richdocument import RichDocument

# do_ocr=DO_OCR (False) keeps docling from loading its OCR models. On these
# documents OCR extracts the same text but permanently costs this kernel
# ~1.9 GiB, since the OCR weights are never released.
construction_plans = RichDocument.from_document_file(
    "construction_docs/construction_plans.pdf", do_ocr=DO_OCR
)
print(construction_plans.get_tables()[0].to_markdown())
print(construction_plans.get_tables()[1].to_markdown())

# Docling's layout and table models are dropped when the converter goes out of
# scope, but only a GC pass actually returns the memory.
reclaim("after parsing construction plans")
```

We need to extract only the tables that are part of the bill of materials, and put those tables into a computer-readable format.

```python
from typing import Literal
from mellea.core import ModelOutputThunk
from mellea.stdlib.components.docs.richdocument import Table
from mellea.stdlib.requirements import req, simple_validate
import pydantic

m = mellea.start_session(model_options=OLLAMA_MODEL_OPTS)

class BOMEntry(pydantic.BaseModel):
    item: str
    quantity: int | str
    notes: str
    category: Literal["lumber", "windows", "doors", "other"]

class BOM(pydantic.BaseModel):
    items: list[BOMEntry]

def _bom_entry_is_well_formed(entry: BOMEntry) -> bool:
    """Checks that the BOMEntry quantity is either an integer or 'allowance'."""
    try:
        int(entry.quantity)
        return True
    except ValueError as e:
        if entry.quantity.lower() == "allowance":
            return True
    return False

def _bom_entries_are_well_formed(s: str) -> bool:
    try:
        bom = BOM.model_validate_json(s)
        return all([_bom_entry_is_well_formed(entry) for entry in bom.items])
    except pydantic.ValidationError as e:
        print(f"Failed on table: {s}")
        return False

# Filter out tables that are not lists of construction items.
@mellea.generative
def is_material_list(table_markdown: str) -> Literal["yes", "no"]:
    """Determines if the table contains a list of construction items. Yes means it contains construction materials."""

async def extract_bom(doc: RichDocument):
    bom_routines = list()
    # Fire off async requests for each table.
    for table in doc.get_tables():
        if is_material_list(m, table_markdown=table.to_markdown()) == "yes":
            print("Reformatting table.")
            next_sub_bom = m.ainstruct(
                "Reformat this table to have four columns: item, quantity, type, and notes (optional).",
                grounding_context={'table': table.to_markdown()},
                requirements=[
                    req(
                        "Quantity row should only contain an integer or Allowance",
                        validation_fn=simple_validate(_bom_entries_are_well_formed)
                    ),
                    req(
                        "type should be one of: lumber, windows, doors, other",
                        validation_fn=simple_validate(lambda x: True)
                    ), # note: this is enforced by the Literal type so no check is required.
                ],
                format=BOM
            )
            bom_routines.append(next_sub_bom)
    
    # wait for all of the async work to finish, then concatenate the results.
    bom_thunks: list[ModelOutputThunk] = [await bom_routine for bom_routine in bom_routines]
    boms = [BOM.model_validate_json(await bom_thunk.avalue()) for bom_thunk in bom_thunks]
    
    # Concatente all of the indiviual BOMs into one large list.
    all_items = []
    for bom in boms:
        all_items.extend(bom.items)
    full_bom = BOM(items=all_items)
    return full_bom

bom = await extract_bom(doc=construction_plans)
reclaim("after extracting the BOM")
```

### Step 2: Find Prices for Materials

LLMs make errors confidently. If we ask a model to generate prices for our bill of materials, the model will either hallucinate or execute a web search. The web search path may appear to be a panacea, but in practice runs into many issues -- especially in B2B settings.

A better solution is to ensure that answers are grounded in context. We can do this by leveraging the Granite RAG Intrinsics library, which are special-build model adapters for **calibrated** context relevance and answerability checking.

Let's assume that our builder has negotiated rates from lumber, windows, and doors -- they want to use those rates for these items. We will demonstrate this basic concept by adding pricing to three components of the procurement list.

To continue from here, we'll need to start a new session using the **huggingface** backend -- adapters are a feature that are not yet supported by the ollama backend.

We're ready to get back to our case study and see how adapter functions can be used in a generative program.

```python
class UnitPriceResponseFmt(pydantic.BaseModel):
    unit_price: float

class TotalPriceResponseFmt(pydantic.BaseModel):
    total_price: float

class BomEntryWithPrice(pydantic.BaseModel):
    item: str
    quantity: int | str
    notes: str
    category: Literal["lumber", "windows", "doors", "other"]
    unit_price: float | None
    total_price: float | None

def get_prices(m: mellea.MelleaSession, bom: BOM) -> list[BomEntryWithPrice]: 
    prices: list[BomEntryWithPrice] = list()
    # Go through every entry in the BOM and figure out if there's an entry in the documents.
    for i, entry in enumerate(bom.items):
        print(f"{i}/{len(bom.items) -1} ({entry.category})")
        # Each iteration feeds a whole catalog through the model. Sweep
        # periodically so the allocator pool does not creep up over the loop.
        if i and i % 10 == 0:
            reclaim(f"pricing loop, item {i}")
        catalog = None
        if entry.category == "windows":
            catalog = windows_doc
        elif entry.category == "doors":
            pass # uncomment to skip
            catalog = doors_doc
        elif entry.category == "lumber":
            # pass # uncomment to skip
            catalog = lumber_doc
        else:
            pass 
        
        if catalog:
            answerability_score = check_answerability(
                f"What is the price of {entry.item}?", 
                documents=[catalog], 
                context=ChatContext(),
                backend=m_hf.backend # type: ignore
            )
            if answerability_score == 'answerable':
                unit_price_response = m.instruct(
                    f"Find the `unit_price` of {entry.item} in the catalog.",
                    grounding_context={"catalog": catalog.text},
                    format=UnitPriceResponseFmt,
                )

                # Excercise: Add additional checks to ensure unit and total price are positive.
                unit_price = UnitPriceResponseFmt.model_validate_json(unit_price_response.value).unit_price
                
                total_price_respsone = m.instruct(f"Find the `total_price` given the `unit_price` and specified `quantity` for {entry.item}",
                                         grounding_context={
                                             "unit_price": str(unit_price),
                                             "quantity": str(entry.quantity)
                                         },
                                         format=TotalPriceResponseFmt)
                total_price = TotalPriceResponseFmt.model_validate_json(total_price_respsone.value).total_price
                
                prices.append(BomEntryWithPrice(
                    item=entry.item,
                    quantity=entry.quantity,
                    notes=entry.notes,
                    category=entry.category,
                    unit_price=unit_price,
                    total_price=total_price,
                ))
                continue
        
        # If we made it here then we did not succeed in adding this item's price from the catalog.
        # Exercise: use web_search tool and citation intrinsic to find prices and validate groundedness.
        # For now, we'll add back the original item with unknown prices.
        prices.append(BomEntryWithPrice(
            item=entry.item,
            quantity=entry.quantity,
            notes=entry.notes,
            category=entry.category,
            unit_price=None,
            total_price=None,
        ))
    return prices

prices = get_prices(m, bom)
reclaim("after pricing")
```

```python
from mellea.backends.model_options import ModelOption
from mellea.stdlib.tools import local_code_interpreter
from mellea.backends.tools import MelleaTool


m = mellea.start_session(
    backend_name="ollama", model_id=REPORT_MODEL, model_options=OLLAMA_MODEL_OPTS
)
mem(f"{REPORT_MODEL} resident")

report_grounding_context = {
    x.item: json.dumps({
        "total_price": x.total_price if x.total_price is not None else "unknown", 
        "category": x.category
    })
    for x in prices
}


pie_chart_result = m.instruct("Use the code interpreter tool to create a pie chart of known cost breakdowns by category. Put the pie chart in /tmp/chart.png", 
           grounding_context=report_grounding_context, tool_calls=True, # type: ignore
           model_options={ModelOption.TOOLS: [MelleaTool.from_callable(local_code_interpreter)]})

pie_chart_result.tool_calls['local_code_interpreter'].call_func() # type: ignore

report = m.instruct(
    "Write an HTML report with a top-line cost breakdown by category and a line-item material list with prices. At the top include the /tmp/chart.png image.", 
    grounding_context=report_grounding_context # type: ignore
)

open("/tmp/report.html", "w").write(report.value)
mem("after building the report")
```
